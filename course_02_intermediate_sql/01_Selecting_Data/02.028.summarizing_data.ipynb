# Summarizing Data
- When analyzing data, we often want to understand the dataset as a while in addition to looking at individual records
- One way to do this, is to summarize the data using SQL aggregate functions
- An aggregate function performs a calculation on several values and returns a single value
- We already know one aggregate function, COUNT()

## Aggregate Functions
- AVG(), SUM(), MIN(), MAX(), COUNT()
- These aggregate functions come after SELECT, exactly like COUNT()

```python
SELECT AVG(budget)
FROM films;
```

This first query gives us the average value from the bidget field of the films table

```python
SELECT SUM(budget)
FROM films;
```

The SUM() function returns the result of adding the values in the budget field

```python
SELECT MIN(budget)
FROM films;
```

The MIN() function returns the lowest budget

```python
SELECT MAX(budget)
FROM films;
```

The MAX() function returns the highest budget.


Note that we operate on the field (or column) with all of these aggregate functions, not the records (or rows)

## Non-numerical data
- Although these functions appear to be mathematical, we can use several of them with both numerical and non-numerical fields

|Numerical fields only|Various data types|
|---|---|
|AVG()|COUNT()|
|SUM()|MIN()|
| |MAX()|

- COUNT() can provide a total of any non-missing, or not null, records in a field regardless of their type
- MIN() and MAX() will give the record that is relatively the lowest or highest
    - Lowest can mean the letter A when dealing with strings 
    - Or the earliest date when dealing with dates
    - The highest or lowest number

## Aliasing when summarizing
- Notice how all query results have automatically updated the field name to the function
- It's best practice to use an alias when summarizing data so that our results are clear to anyone reading our code


!["Aliasing when summarizing"](<aliasing_when_summarizing.png>)

```python
SELECT MIN(country) AS min_country
FROM films;
```

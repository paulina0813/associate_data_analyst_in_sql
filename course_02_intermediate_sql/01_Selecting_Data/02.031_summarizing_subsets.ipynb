# Summarizing Subsets

## Using WHERE with aggregate functions
- We can combine aggregate functions with the WHERE clause to gain further insights from our data
- That's because the WHERE clause executes before the SELECT statement
- For example, to get the average budget of movies made in 2010 or later, we would select the average of the budget field from the films table where the release year is greater than or equal to 2010.

```python
SELECT AVG(budget) AS avg_budget
FROM films
WHERE release_year >= 2010;
```

Here are a few more examples using the other functions:
- We find the total budget of movies made in 2010 using the SUM() function
- Next, we get the smallest budget using the MIN() function

```python
SELECT SUM(budget) AS sum_budget
FROM films
WHERE release_year = 2010;
```

```python
SELECT MIN(budget) AS min_budget
FROM films
WHERE release_year = 2010;
```

- Then, we query the highest budget using the MAX() function

```python
SELECT MAX(budget) AS max_budget
FROM films
WHERE release_year = 2010;
```

Finally, we query the count of the number of budgets using the COUNT function, which gives us the total number of non-missing values in the budget field

```python
SELECT COUNT(budget) AS count_budget
FROM films
WHERE release_year = 2010;
```

Now that we are doing different operations with our numerical values, there will likely be some crazy decimals that we'll want to clean

## ROUND()
- In SQL, we can use ROUND() to round our number to a specified decimal
- There are 2 parameters for round:
    - The number we want to round
    - The decimal place we want to round to
- ROUND(number_to_round, decimal_places)
- Can only be used with numerical fields

!["Round Function"](<round_function.png>)

## ROUND() to a whole number
- The second parameter in our ROUND() function is optional, so we can leave it out if we want to round to a whole number
- We would get the same result if we passed zero as the second argument

!["Round to Whole Number"](<round_to_whole_number.png>)


## ROUND() as a negative parameter
- We could also pass a negative number as the second parameter and still get a result
- Here, a function is rounding to the left of a decimal point instead of to the right
- Using -5 as the decimal place parameter will cause the function to round to the hundred thousand or five places to the left

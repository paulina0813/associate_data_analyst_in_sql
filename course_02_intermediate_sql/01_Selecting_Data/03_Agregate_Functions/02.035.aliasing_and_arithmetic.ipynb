# Aliasing and Arithmetic

## Arithmetic
- We can perform basic arithmetic symbols like +, -, *, and /
- Using parenthesis with arithmetic indicates to the processor when the calculation needs to execute
- We can add, subtract, multiply, and divide as follows

```python
SELECT (4+3);
```

```python
SELECT (4 * 3);
```

```python
SELECT (4 - 3);
```

```python
SELECT (4 / 3);
```

In these examples, the parenthesis are not required as only one calculation takes place; however, they provide more clarity to the code.

But, the division gives a result of 1; why is that?

Similar to other programming lagauges, SQL assumes we want to get an integer back if we divide an integer by an integer

When dividing, we can add decimal places to our numbers if we want more precision.

For example, the following query gives us the result we would expect: 1.33333...

```python
SELECT (4.0 / 3.0);
```

## Order of execution
- Step 1: FROM
- Step 2: WHERE
- Steo 3: SELECT (aliases are defined here)
- Step 4: LIMIT

- Aliases defined in the SELECT clause cannot be used in the WHERE clause due to the order of execution
- Since the alias will not have been created by the time the WHERE statement is executed, the code will return an error

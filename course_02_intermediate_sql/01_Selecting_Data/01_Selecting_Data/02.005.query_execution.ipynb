# Query execution

## Order of execution
- SQL is not processed in its written order
- Before any data can be selected, the table from which the data will be selected needs to be indicated
- Next, out selection is made
- Finally, our results are refined
    - Here, we use the LIMIT keyword that limits the results to a sepcified number of records
    - LIMIT limits how many results we return
- Good to know the processing order for debugging and aliasing
- Suppose we need to refer to an alias later on un our code. In that case, that alias will only make sense to a processor when its declaration in the SELECT statement is processed before the alias reference is made anywhere else in the query 
- Aliases are declared in the SELECT statement

```python
-- Order of execution
SELECT name
FROM people
LIMIT 10;
```

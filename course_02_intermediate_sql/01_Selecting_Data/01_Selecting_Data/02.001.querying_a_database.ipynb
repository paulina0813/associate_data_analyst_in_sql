# Querying a Database

## COUNT()
- COUNT()
- Counts the number of records with a value in a field
- For example, to count the number of birth dates present in the people table, we will use

```python
SELECT COUNT(birthdate) AS count_birthdates
FROM people;
```

## COUNT() multiple fields
Here, we are counting both the number of names and birth dates present in the people table

```python
SELECT 
    COUNT(name) AS count_names, 
    COUNT(birthdate) AS count_birthdates
FROM people;
```

## Using * with COUNT()
- COUNT(field_name) counts values in a field
- COUNT(*) counts records in a table
- \* represents all fields
For example, this code gives the total number of records in the people table


```python
SELECT COUNT(*) AS total_records
FROM people;
```

## DISTINCT
Our results will often include duplicates. We can use the DISTINCT keyword to select all the unique values from a field
- DISTINCT removes duplicates to return only unique values
- This will be useful if we are interested in knowing which languages are represented in our films table?

```python
SELECT language
FROM films;
```

Adding DISTINCT to out query will remove all duplicates

```python
SELECT DISTINCT(language)
FROM films;
```

## COUNT() with DISTINCT
- Combine COUNT() with DISTINCT to count unique values

This query counts the number of distinct birth dates in the people table

```python
SELECT COUNT(DISTINCT birthdate) AS count_distinct_birthdates
FROM people;
```

- COUNT() includes duplicates
- DISTINCT excludes duplicates

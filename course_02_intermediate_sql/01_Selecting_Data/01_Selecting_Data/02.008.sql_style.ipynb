# SQL Style

## SQL formatting
- Formatting is not required 
- New lines, capitalization, and indentation are not required in SQL

The following code will run just fine, returning the first three titles, release yearsm and countries from the films table.

```python
select title, release_year, country from films limit 3;
```

However, writing queries like this won't make us any friends in the SQL world because the lack of formatting makes the code difficult to read, especially as queries become more complex

## Best practices
Over time, SQL users have developed style standards that are generally accespted across industries. This code returns the same result as the code above, but it is much easier to read due to the addition of capitalized keywords and new lines between them

```python
SELECT title, release_year, country
FROM films
LIMIT 3;
```

- Capitalize keywords
- Add new lines

## Style guides
While keyword capitalization and new lines are standard practice, many of the finer detailes of SQL style are not.

For instance, some SQL users prefer to create a new line and indent each selected field when a query selects multiple fields, as the following query

```python
SELECT 
    title, 
    release_year, 
    country
FROM films
LIMIT 3;
```

Because of the different formatting styles, it's helpful to follow a SQL stye guide, such as Holywell's, which outlines a standard of best practices for indentation, capitalization, and naming conventions for tables, fields, and aliases.

https://www.sqlstyle.guide/


## Semicolon
Like capitalization and new lines, this semicolon is unnecessary in Postgre SQL, we could leave it out of the query and still expect the dame results with no errors.

However, including a semicolon at the end of the query is considered best practice for several reasons.

- Some SQL flavors require ir, so it's a good habit to have
- Including a semicolon in a PostgreSQL query means that the query is more easily translated to another flavor if necessary.

- A semicolon at the end of a query indicates its end, which is helpful in a file containing several queries

## Dealing with non-standard field names
- When creating a table, a SQL mistake is including spaces in a field name
    - release year instead of release_year
- To query that table, we'll need to enclose that field name in double quotes to indicate that, despite being two words, the name refers to just one field
- Put non-standard field names in double-quotes

For example, if a sloppy SQL coder had named a field release-space-year as two words, we'll need to update the query we've seen to this:

```python
SELECT title, "release year", country
FROM films
LIMIT 3;
```

# Sorting Results
- Put the data in a specific order

## ORDER BY
- This keyword is used to sort results of one or more fields
- When used on its own, it's written after the FROM statement

```python
SELECT title, budget
FROM films
ORDER BY budget;
```

```python
SELECT title, budget
FROM films
ORDER BY title;
```

- ORDER BY will sort in ascending order by default
    - From smallest to biggest or form A-Z
- The queries above sort the budgets from smallest to biggest and the titles alphabetically
- Our database contains titles that start with symbols and numbers, these come before the letter A
- We could also add the ASC keyword to our query to clarify that we are sorting in ascending order; the results are the same and out code is more readable

```python
SELECT title, budget
FROM films
ORDER BY budget ASC;
```

### DESCending
- We can use the DESC keyword to sort the results in descending order.
- The following query gives us the film titles sorted by budget from biggest to smallest

```python
SELECT title, budget
FROM films
ORDER BY budget DESC;
```

However, our data contains a lot of null values, which will be at the top.

We can add a WHERE clause before ORDER BY to filter the budget field for only non-null values and improve our results.

```python
SELECT title, budget
FROM films
WHERE budget IS NOT NULL
ORDER BY budget DESC;
```

## ORDER BY multiple fields
- ORDER BY field_one, field_two
- It will sort by the first field specified, then sort by the next, etc
- The second field we sort by can se thought of as a tie breaker when the first field is not decisive in telling the order

!["ORDER BY multiple fields"](<order_by_multiple_fields.png>)

When we sort by the first field, we can see that there is a tie, so we sort by a second field to break that tie.

## Different orders
- We can also select a different order for each field we are sorting

```python
SELECT birthdate, name
FROM people
ORDER BY birthdate, name DESC;
```

- For example, here we are sorting birthdate un ascending order and name in descending order
- ORDER BY falls towards the end of the order of execution we already know, coming in just before LIMIT

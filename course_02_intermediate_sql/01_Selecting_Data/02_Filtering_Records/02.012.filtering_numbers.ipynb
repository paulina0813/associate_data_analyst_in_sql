# Filtering Numbers

## WHERE
- It allows us to focus on only the data that is relevant to our business questions
- WHERE filtering clause
- To filter numbers, we will be using comparison operators like greater than

```python
SELECT title
FROM films
WHERE release_year > 1960;
```

Here is an example of a query where we filtered to see only films released after the year 1960 using the greater than operator.

## Comparison operators
We would use the less-than operator to see films released before the year 1960

```python
SELECT title
FROM films
WHERE release_year < 1960;
```

We would use the less-than or equal to operator to see films released during or before 1960

```python
SELECT title
FROM films
WHERE release_year <= 1960;
```

If we want to see films released in a specific year, we can use equals

```python
SELECT title
FROM films
WHERE release_year = 1960;
```

If we wanted to filter films to see all releases EXCEPT thos from the year 1960, we would combine the less than and greater than symbols as shown here

```python
SELECT title
FROM films
WHERE release_year <> 1960;
```

- \> greater than or after
- < less than or before
- = equal to
- \>= greater than or equal to
- <= less than or equal to
- <> not equal to 

## WHERE with strings
- WHERE and the comparison operator, equals, can also be used with strings
- In these cases, we will have to use single quotation marks around the strings we want to filter

For example, here, we want to filter titles where the country is Japan

```python
SELECT title
FROM films
WHERE country = 'Japan';
```

## Order of execution
Similar to LIMIT, this clause comes after the FROM statement when writing a query. If we were to use both WHERE and LIMIT, the written order will be
- SELECT
- FROM
- WHERE
- LIMIT

However, the order of execution will be:
- FROM
- WHERE
- SELECT
- LIMIT

```python
-- Written code
SELECT item
FROM coats
WHERE color = 'green'
LIMIT 5;
```

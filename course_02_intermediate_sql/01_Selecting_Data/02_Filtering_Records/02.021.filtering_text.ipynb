# Filtering text
- We can use the WHERE clause to filter text data
- So far, we've only been able to filter by specifying the exact text we're interested in
- We'll often want to search for a pattern rather than a specific text tring
- We'll be introducing 3 more SQL keywords into our vocabulary to help us achieve this
    - LIKE
    - NOT LIKE
    - IN

    ## LIKE
    - Used to search for a pattern in a field
    - We use a wildcard for some other values to accomplish this.
    - There are two wildcards with LIKE, the percent and the underscore
        - % - match zero, one, or many characters
        - _ - match a single character

```python
SELECT name
FROM people
WHERE name LIKE 'Ade%';
```

    The query above matches people like Adel, Adelaide, and Aden.


    !["Percent wildcard"](<percent_wildcard.png>)


    The underscore wildcard will match a single character

```python
SELECT name
FROM people
WHERE name LIKE 'Ev_';
```

    The query above matches only three-letter names like Eve


    !["Underscore wildcard"](<underscore_wildcard.png>)


   -  We'd also see names like Eva if it were in our dataset. Eva Mendes, however, would not be visible.

    ## NOT LIKE
    - Used to find records that don't match the specified pattern.
    - In this query, we are finding records for people who do not have A. as part of their first name

```python
SELECT name
FROM people
WHERE name NOT LIKE 'A.%';
```

It's important to note that this operation is case-sensitive, so we must be mindful of what we are querying

## Wildcard position
We can find values that start, end, or contain characters in any position, as well as find records of a certain length.

For example, this code will find all people whose name ends in r

```python
SELECT name
FROM people
WHERE name LIKE '%r';
```

This code will find records where the third character is t

```python
SELECT name
FROM people
WHERE name LIKE '__t%';
```

## WHERE, OR
If we wanted to filter based on many conditions or a range of number, we could chain several ORs to the WHERE clause based on what we know, but that can get messy

```python
SELECT title
FROM films
WHERE release_year = 1920,
    OR release_year = 1930,
    OR release_year = 1940;
```

## WHERE, IN
- A helpful operator in the scenario above is IN.
- The IN operator allows us to specify multiple values in a WHERE clause, making it easier to set numerous OR conditions

```python
SELECT title
FROM films
WHERE release_year IN (1920, 1930, 1940);
```

 - This can also be used in a text field
 - For example, if we want to find the title WHERE the associated country is either Germany or France, we could use

```python
SELECT title
FROM films
WHERE country IN ('Germany', 'France');
```

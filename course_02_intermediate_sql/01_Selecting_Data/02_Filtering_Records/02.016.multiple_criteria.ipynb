# Multiple criteria

There will often be more than one criteria that we want to filter on

There are 3 additional keywords that will allow us to enhance our filters when using WHERE by adding multiple criteria
- AND
- OR
- BETWEEN

In the context of coats, we could look at coats where the color is yellow OR the length is short 

```python
SELECT *
FROM coats
WHERE color = 'yellow' OR length = 'short';
```

Or we could filter coats where both criteria are true

```python
SELECT *
FROM coats
WHERE color = 'yellow' AND length = 'short';
```

We could also look for coats hat have between one and five buttons

```python
SELECT *
FROM coats
WHERE buttons BETWEEN 1 AND 5
```

## OR operator
- Use OR when you need to satisfy at least one condition
- In SQL, we combine OR with WHERE to achieve this type of filtering

```python
-- Correct
SELECT title
FROM films
WHERE release_year = 1994
    OR release_year = 2000;
```

The query above returns all films released in either 1994 or 2000.

We must specify the field for every OR condition, so the query below is invalid. This query hasn't specified what field or operator should be associated with the year 2000

```python
-- Invalid
SELECT title
FROM films
WHERE release_year = 1994 OR 2000;
```

## AND operator
- Use AND if we need to satisfy all criteria
- In SQL, we combine AND with WHERE to achieve this type of filtering

```python
-- Correct
SELECT title
FROM films
WHERE release_year > 1994
    AND release_year < 2000;
```

The query above gives us all titles of films released between 1994 and 2000. We must specify the field for every AND condition, so the query below is invalid. This query hasn't specified what field should be associated with the year 2000

```python
-- Invalid
SELECT title
FROM films
WHERE release_year > 1994 AND < 2000;
```

## AND, OR
- Filter films released in 1994 or 1995, and certified PG or R
- We can combine AND and OR to answer this question

If a query has multiple filtering conditions, we will need to enclose the individual clauses in parenthesis to ensure the correct execution order; otherwisee, we may not get the expected results

```python
SELECT title
FROM films
WHERE (release_year = 1994 OR release_year = 1995) 
    AND (certification = 'PG' OR certification = 'R');
```

## BETWEEN, AND
As we've learned can use this query to get titles of all films released in and between 1994 and 2000

```python
SELECT title
FROM films
WHERE release_year >= 1994
    AND release_year <= 2000;
```

Checking for ranges like this is very common, so in SQL the BETWEEN keyword provides a valuable shorthand for filtering values within a specified range

```python
SELECT title
FROM films
WHERE release_year BETWEEN 1994 AND 2000;
```

It's important to remember that BETWEEEN is inclusing, meaning the results contain the beginning and end values

## BETWEEN, AND, OR
Like the WHERE clause, the BETWEEN clause can be used with multiple AND and OR operators, so we can build up our queries and make them even more powerful.

For example, we can get the titles of all films released between 1994 and 2000 from the United Kingdom

```python
SELECT title
FROM films
WHERE release_year BETWEEN 1994 AND 2000 AND country = 'UK';
```

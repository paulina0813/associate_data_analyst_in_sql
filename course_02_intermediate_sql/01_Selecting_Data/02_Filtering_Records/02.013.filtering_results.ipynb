# Filtering results

The WHERE clause allows you to filter based on text and numeric values in a table using comparison operators.

What does the following query return?

```python
SELECT title
FROM films
WHERE release_year > 2000;
```

## Possible Answers
1. Films released before the year 2000
2. Films released after the year 2000
3. Films released after the year 2001
4. Films released in 2000

## Correct Answer
2. Films released after the year 2000

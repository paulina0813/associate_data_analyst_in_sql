# NULL values

## Missing values
- COUNT(field_name) includes only non-missing values
- COUNT(*) includes missing values
- NULL representa a missing or unknown values
    - Missing values:
        - Human error
        - Information not available
        - Unknown
    - Knowing how to handle these fields is essential as thet can affect any abalyses we do

## IS NULL
One quick way to see how much of our data is missing, is by using IS NULL with the WHERE clause.

Here is an example where we have checked to see which names fo not have a recorded birthdate in our table. 

```python
SELECT name
FROM people
WHERE birthdate IS NULL;
```

This is an example of missing birthdates in the people

```python
SELECT COUNT(*) AS no_birthdates
FROM people
WHERE birthdate IS NULL;
```

## IS NOT NULL
- If we want to filter out missing values, so we only get results that are NOT NULL, we can use the IS NOT NULL operator
- For example, this query gives the count of all people whose birth dates are not missing in the people table

```python
SELECT COUNT(birthdate) AS no_birthdates
FROM people
WHERE birthdate IS NOT NULL;
```

## COUNT() vs IS NOT NULL
There might be a question about the difference between using COUNT with a field name and using the same COUNT with the added WHERE clause with IS NOT NULL. 

The answer is there is no difference, as both will be counting non-missing values

```python
SELECT
    COUNT(certification) AS count_certification
FROM films;
```

```python
SELECT
    COUNT(certification) AS count_certification
FROM films
WHERE certification IS NOT NULL;
```

# Grouping data

- SQL allows us to group with the GROUP BY clause

    ## GROUP BY single fields
    - It is commonly used with aggregate functions to provide summary statistics, particularly when only grouping a single field, certification, and selecting multiple fields, certification and title
    - The aggregate function will reduce the non-grouped fields to one record only, which will need to correspond to one group
    - For example, if we want to see how many films correspond to each rating

```python
SELECT certification, COUNT(title) AS title_count
FROM films
GROUP BY certification;
```

- SQL will return an ERROR if we try to SELECT a field that is not in our GROUP BY clause
    - We'll need to correct this by adding an aggregate function around title

## GROUP BY multiple fields
- We can use GROUP BY multiple fields, similar to ORDER BY
- The order in which we write the fields affects how the data is grouped
- The query here selects and groups certification and language while aggregating the title

```python
SELECT certification, language, COUNT(title) AS title_count
FROM films
GROUP BY certification, language;
```

- The result shows that we have 5 films that have missing values for both certification and language, 2 films that are Unrated and in Japanese, two films that are rated R and in Norwegian, and so on.


!["GROUP BY multiple fields"](<group_by_multiple_fields.png>)


## GROUP BY with ORDER BY
- We can combine GROUP BY with ORDER BY to group our results, make a calculation, and then order our results
- For example, we can clean up one of our previous queries by sorting the results by the title count in descending order

```python
SELECT certification, COUNT(title) AS title_count
FROM films
GROUP BY certification
ORDER BY title_count DESC;
```


!["Ordered GROUP BY"](<ordered_group_by.png>)


- ORDER BY is always written after GROUP BY, and notice that we can refer back to the alias within the query. That is because of the order of execution

### Order of Written Code
- SELECT
- FROM
- GROUP BY
- ORDER BY
- LIMIT

### Order of execution
- FROM
- GROUP BY
- SELECT
- ORDER BY
- LIMIT

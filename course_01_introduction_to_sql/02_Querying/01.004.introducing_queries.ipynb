## Keywords
- SELECT - this indicates which fields should be selected
- FROM - this indicates the table in which these fields are located

```python
-- Query
SELECT name
FROM patrons;
```

In this case we can to view the data from the names field in the patrons table.

We end the query with a semicolon to indicate that the query is complete.

Keywords are capitalized while table names and fields are all lowercase

The result will list all patron names


!["Query Example 01"](<query_example_01.png>)


To select multiple fields, we can list multiple field names after the SELECT keyword, separated by commas

```python
-- Query with multiple fields
SELECT card_num, name
FROM patrons;
```

In this example we will select card number and name from our patrons table, and the results will show the field names in the same order that they are in the query



!["Query Example 02"](<query_example_02.png>)


We can also select theree field names by listing all 3 field names


```python
-- Query with 3 fields
SELECT name, card_num, total_fine
FROM patrons;
```

!["Query Example 03"](<query_example_03.png>)

If we want to select all of the fields in one table, we could select each individual field after the SELECT statement, but there's an easier way.

We can tell SQL to select all fields using as asterisk or star, also known as a wildcard character instead.

```python
SELECT *
FROM patrons;
```

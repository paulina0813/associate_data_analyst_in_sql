## Aliasing
- Renaming columns for clarity or brevity


For instance, let's consider an employees table that includes a fueld for each employee's name and their hire date

!["Employees Table"](<employees_table.png>)

Because the table only contains each employee's first name, we can use the AS keyword to alias the name field as first_name


```python
SELECT name AS fist_name
FROM employees;
```

This changes the field name to first_name in the result set, while the actual field name in the table remains as name.

!["Aliasing Example 01](<aliasting_example_01.png>)


Some SQL questions require a way to resturn a list of unique values.

Let's imagine we want ti create a list of years when we hired our employees

```python
SELECT year_hired
FROM employees;
```

Selecting the year_hired field from the employees table shows some years multiple times, which is not what we want.

To get a list of years with no repeat values, we can add the DISTINCT keyword before the year_hired field name in the SELECT statement.

```python
SELECT DISTINCT(year_hired)
FROM employees;
```

!["Unique Values"](<unique_values_01.png>)

Now, we can see that all of out employees were hired in just four years.

It's possible to return the unique combinations of multiple field values by listing multiple fields after the DISTINCT keyword.

Let's say we want to know the years that different departments hired employees.

```python
SELECT dept_id, years_hired
FROM employees;
```

!["Years Hired by Department"](<years_hired_by_department.png>)

We could use this SQL query to look at this information, selecting the dept_id and year_hired from the employees table.

Looking at the results, we see that department three hired two employees in 2021.

To avoid repeating this information, we could add the DISTINCT keyword before the fields to SELECT

```python
SELECT DISTINCT(dept_id), year_hired
FROM employees;
```

!["Unique Values"](<unique_values_02.png>)

Notice that the depatrtment id and year_hired fields still have repeat values individually but none of the records are the same: they are all unique combinations of the two fields.

Finally, let's explore saving SQL result sets.

A view is a saved SQL query that acts like a virtual table. Views don't store data, they store the query, ensuring the results are always up-to-date with the latest database changes. To create a view, use CREATE VIEW, followed by the view name, and AS to define the query.

```python
CREATE VIEW employee_hire_years AS
SELECT id, name, year_hired
FROM employees;
```

employee_hire_years can be created by saving a query that selects specific fields from the employees table.

Note that creating a view doesn't produce a resuly set, it only saves the query for reuse.

Once a view is created, we can query it just as we would by selecting FROM the view

# Review UNION and INTERSECT
Which of the following definitions of set operations is correct?

## Possible Answers
1. UNION: returns all records (potentially duplicates) in both tables
2. UNION ALL: returns only unique records
3. INTERSECT: returns only records appearing in both tables
4. None of the above definitions are correct.


## Correct Answer
3. INTERSECT: returns only records appearing in both tables

## UNION and UNION ALL syntax
- We perform a SELECT statement on our first table, a SELECT statement on our second table, and specify a set operation (in this example, either UNION or UNION ALL) between them

```python
-- UNION syntax

SELECT *
FROM left_table
UNION
SELECT *
FROM right_table;
```

```python
-- UNION ALL syntax

SELECT *
FROM left_table
UNION ALL
SELECT *
FROM right_table;
```

## JOIN vs Set Operations

- Note that set operations don't require a field to join ON
- This is because hey don't do quite the same thing as the JOIN operations
- Rather than comparing and merging tables on the left and right, thet stack fields on top of one another
- For all set operations, the number of selected columns and their respective data types must be identical
- For instance, we can't stack a number field on top of a character field
- The result will onlu use field names (or aliases, if used) of the first SELECT statement of the query


!["Set Theory 02"](<set_theory_02.png>)


- We now return to our world leaders example
- In this example, we use one more table from our leaders databse: the monarchs table
- We can use a UNION on the monarchs and prime_ministers tables to show all the different prime_ministers and monarchs in these two tables
- Note that the monarch field has been aliased as 'leader'
- The monarch and prime_minister field will be combined under the leader field, even if we only aliased the monarch field

```python
SELECT monarch AS leader, country
FROM monarchs
UNION
SELECT prime_minster, country
FROM prime_ministers
ORDER BY country, leader
LIMIT 10;
```

!["Set Theory 03"](<set_theory_03.png>)

- The UNION returns the top 10 monarchs, prime_ministers, and their corresponding countries
- Recall that the monarchs for Oman and Brunei are also prime ministers.
    - UNION only lists them once each, because they are the same person
- Norway is listed twice because its monarch and prime minister are different people
- A UNION ALL will tell us if there are many monarhcs who also act as prime ministers

```python
SELECT monarch AS leader, country
FROM monarchs
UNION ALL
SELECT prime_minster, country
FROM prime_ministers
ORDER BY country, leader
LIMIT 10;
```

!["Set Theory 04"](<set_theory_04.png>)

- After UNION ALL, Brunei and Oman are now listed twice

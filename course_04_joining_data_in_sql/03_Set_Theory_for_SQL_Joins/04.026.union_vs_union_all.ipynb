# UNION vs. UNION ALL

Nice work learning all about UNION and UNION ALL!

Two tables, languages and currencies, are provided. Run the queries provided in the console and select the correct answer for the multiple-choice questions in this exercise.

## Instructions
1. What result will the following SQL query produce?

```python
SELECT * 
FROM languages
UNION
SELECT * 
FROM currencies;
```

### Possible Answers
1. All records from both tables, dropping duplicate records (if any)
2. A SQL error, because languages and currencies do not have the same number of fields
A SQL error, because languages and currencies do not have the same number of records


### Correct Answer
2. A SQL error, because languages and currencies do not have the same number of fields


## Instructions
2. What result will the following SQL query produce?

```python
SELECT code FROM
languages
UNION ALL
SELECT code FROM 
currencies;
```

### Possible Answers
1. An ordered list of each country code in languages and currencies, including duplicates
2. An ordered list of each unique country code in languages and currencies
3. An unordered list of each country code in languages and currencies, including duplicates
4. An unordered list of each unique country code in languages and currencies

### Correct Answer
3. An unordered list of each country code in languages and currencies, including duplicates

## Instructions
3. What will the following SQL query produce?


```python
SELECT code 
FROM languages
UNION
SELECT curr_id 
FROM currencies;
```

### Possible Answers
1. An empty result
2. A stacked list of every curr_id from currencies and every code from languages
3. A SQL error, because code and curr_id are not of the same data type
4. A SQL error, because code and curr_id do not have the same name


### Correct Answer
3. A SQL error, because code and curr_id are not of the same data type

# At the INTERSECT

- INTERSECT takes two tables as input, and returns only the records that exist in both tables.


!["INTERSECT diagram 01"](<intersect_diagram_01.png>)


!["INTERSECT diagram 02"](<intersect_diagram_02.png>)


- In the diagram shown, we have two tables, left_table and right_table
- The result of performing INTERSECT on these tables is only the records common to both tables: the first record
- All records that are not of interest to the INTERSECT operation are faded out


## INTERSECT syntax
- The syntax for this set operation is very similar to that of UNION and UNION ALL
- We perform a SELECT statement on out first table, a SELECT statement on out second table, and sepcify out set oeprator between them

```python
SELECT id, val
FROM left_table
INTERSECT
SELECT id, val
FROM right_table;
```

## INTERSECT vs INNER JOIN on two columns

- Let's compare an INTERSECT to performing an INNER JOIN on two fields with identical field names
- Similar tu UNION, for a record to be returned, INTERSECT requires all fields to match, since in set operations we do not specify any fields to match on
- This is also why it requieres the left and right table ti have the same number of columns in order to compare records

```python
-- INTERSECT syntax
SELECT *
FROM left_table
INTERSECT
SELECT *
FROM right_table;
```

- In the figure shown, only records where both columns match are returned


!["INTERSECT diagram 02"](<intersect_diagram_02.png>)


- In INNER JOIN, similar to INTERSECT, only results where both fields match are returened

```python
-- INNER JOIN syntax
SELECT *
FROM left_table
INNER JOIN right_table
ON left.id = right.id
    AND left.val = right.val;
```

1["INNER JOIN diagram"](<inner_join_diagram.png>)


- INNER JOIN will return duplicate values, whereas INTERSECT will only return common records once.
- As we know from earlier lessons, INNER JOIN will ad more columns to the result set

- Let's take a look at how we can use INTERSECT to determine all countries that have both a prime minister and a president
- Each SELECT statement in our query has the same number of columns, of identical data tyes, in order for the set operation to be performed

```python
SELECT country AS intersect_ountry
FROM prime_ministers
INTERSECT
SELECT country
FROM presidents;
```

!["INTERSECT results 01"](<intersect_results_01.png>)


- The result of the query is the four countries with both a prime minister and a president in our leaders database
- As with UNION, the result set uses the field names provided for the left_table, whether aliased or not


## INTERSECT on two fields
- If we use two columns for an INTERSECT. the result will be an empty table for the example above


```python
SELECT country, prime_minister AS leader
FROM prime_ministers
INTERSECT
SELECT country, president
FROM presidents;
```

!["INTERSECT results 02"](<intersect_results_02.png>)

- However, recall that there are monarchs in our database who also act as prime_ministers

# Defining relationships

## One-to-many relationships
- Onse single entity can be associated with several entities
- Think about a music library, one artist can produce manu songs over their career
    - The same applies to authors and their books, directors and movie titles, and so on

## One-to-one relationships
- These imply unique pairings between entities and are therefore less common
- An example of a one-to-one relationship: one fingerprint for one finger


!["One to one relationships"](<one_to_one_relationships.png>)


!["One to one relationships 02"](<one_to_one_relationships_02.png>)


Imagine fingerprint scanning at airport border control. Border control might have two tables in a database: a table of people and a table of their prints.

Each person's fingerprint could be merged with the individuals table using their passport number.

Note that although an individual will have four fingerprints stored in a fingerprints table, this is not a one-to-many relationship. It's a one-to-one relationship, because a single individual, represented as one record on the individuals table - will hace one set of fingerprints - represented as one record on the fingerprints table. The four different fingerprints are four different fields in each record.

## Many-to-many relationships
- An example of this is languages and countries

!["Many to many relationships"](<many_to_many_relationships.png>)

Here we show the official languages of Germany, Belgium, and the Netherlands, where we see that many languages can be spoken in many countries.

For example, belgium has 3 official languages: French, German, and Dutch.

Conversely, languages can be official in many languages: Dutch is an official language of both the Netherlands and Belgium, but not Germany

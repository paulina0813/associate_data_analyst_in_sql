## Our first INNER JOIN
- The syntax for performing an INNER JOIN on our tables is shown step by step
- It is common to begin constructing the query with the join first, and selecting the fields later
- After FROM, we list the left table, followed by the INNER JOIN keyword and the right table
- We then specify the field to match the tables on, using the ON keyword
- Here, we use the "country" field
- Lastly, we add SELECT at the start of the query and choose the fields we want returned
- When selecting columns that existe in both tables, such as "country" and "continent"
- The table.column_name format must be used to avoid a SQL error

```python
-- Inner join of presidents and prime_ministers, joining on country
SELECT 
    prime_ministers.country, 
    prime_ministers.continent, 
    prime_minister, 
    president
FROM presidents
INNER JOIN prime_ministers
ON presidents.country = prime_ministers.country;
```

The result of our query is displayed: four countries in our database have both presidents and prime ministers.

## Aliasing

In our INNER JOIN, we've had to type out "president" and "prime_minister" several times. 

Luckily, we can alias table names using the same AS keyword used to alias column names.

Here, we use aliases p1 and p2 in both our SELECT and ON clauses to shorten our query

```python
-- Inner join of presidents and prime_ministers, joining on country
SELECT 
    p2.country, 
    p2.continent, 
    prime_minister, 
    president
FROM presidents AS p1
INNER JOIN prime_ministers AS p2
ON p1.country = p2.country;
```

When joining on two identical column names, we can employ de USING command, followed by the shared column name in parenthesis.

Here, since the join field is named "country" in both tables, we can use USING

```python
-- Inner join of presidents and prime_ministers, joining on country
SELECT 
    p2.country, 
    p2.continent, 
    prime_minister, 
    president
FROM presidents AS p1
INNER JOIN prime_ministers AS p2
USING(country);
```

# Subqueries inside FROM
- Let's say we are interested in all the countries with monarchs along with the most recent country to gain independence in that continent
- A starting SQL query ti pull all the most recent independence years by continent is shown

```python
SELECT continent, MAX(indep_year) AS most_recent
FROM states
GROUP BY continent;
```

- The query groups records by continent and returns the most recent independence year for each group, by using the MAX() function as most_recent
- How do we filter this for continents with monarchs?
- We haven't seen yet that we can iclude multiple tables in a FROM clause by adding a comma between them
- In the SQL syntax shown, we include two different tables, left_table and right_table, in our FROM clause

```python
SELECT left_table.id, left_val
FROM left_table, right_table
WHERE left_table.id = right_table.id
```

- The diagram shown illustrates that this query returns duplicated from multiple occurrences of id when it matches in both tables
- We can drop duplicates using the DISTINCT command in the query shown

```python
SELECT DISTINCT left_table.id, left_val
FROM left_table, right_table
WHERE left_table.id = right_table.id
```

!["Records inside monarchs diagram 02"](<records_inside_monarchs_diagram_02.png>)


- Let's investigate a way to solve our problem using this trick
- We can nest the subquer we initially wrote into a FROM statement, so that we are selecting from both monarchs and our subquery, aliased as "sub"
- We also use the WHERE clause as before to identify records in both tables that match on continent

```python
-- Query to return continents with monarchs and the year the most recent country gained independence

FROM monarchs,
    (SELECT
        continent,
        MAX(indep_year) AS most_recent
    FROM states
    GROUP BY continent) AS sub
WHERE monarchs.continent = sub.continent
```

- As we saw on the previous slide, because the continents from our subquery will find multiple matches in monarchs, this match will  return duplicates
- As before, in our SELECT statement we use the DISTINCT command to drop duplicate continents, and select sub.most_recent to get the most recent independence year for each continent
- Lastly, we order by continent

```python
-- Query to return continents with monarchs and the year the most recent country gained independence
SELECT DISTINCT monarchs.continent, sub.most_recent
FROM monarchs,
    (SELECT
        continent,
        MAX(indep_year) AS most_recent
    FROM states
    GROUP BY continent) AS sub
WHERE monarchs.continent = sub.continent
ORDER BY continent;
```

- There we have it: the continents with monarchs in our database, and the most recent year of independence in those continents
- That's how we include a subquery as a temporary table in our FROM clause and then SELECT from it

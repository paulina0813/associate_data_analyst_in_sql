# Subqueries inside WHERE and SELECT

- All semi joins and anti joins we have seen included a subquery in WHERE
- WHERE is the most common place for subqueries because filtering data is one of the most common data manipulation tasks

## Syntax for query using WHERE IN statement
- Arguments to the IN operator are not limited to lists typed out by us
- We can include a SQL subquery as an argument for the IN operator, provided the result of the subquery is of the same data type as the field we are filtering on.
- The query shown will only work if some_field is of the same data type as some_numeric_field, because the result of the subquery will be a numeric field

```python
SELECT *
FROM some_table
WHERE some_field IN
    (SELECT some_numeric_field
    FROM another_table
    WHERE field2 = some_condition);
```

- Subqueries inside WHERE can be from the same table or from a different table, and here, the subquery is from a different table
- The second most common type of subquery is inside a SELECT clause


## Subqueries inside SELECT
- Say you want to count the number of monarchs listed in the monarchs table for each continent in the states table
- Our subquery requires two things
- First it needs to count all monarchs
- Second, it needs a WHERE statement matching the continent fields in the two tables

```python
SELECT DISTINCT continent,
    (SELECT COUNT(*)
    FROM monarchs
    WHERE states.continent = monarch.continent) AS monarch_count
FROM states;
```

- This subquery follows the selection of DISTINCT continents, and will therefore count all monarchs within them in the SELECT statement
- A subquery inside a SELECT statement like this requieres an alias, like monarch_count here

# Subquerying with semi joins and anti joins

- The six joins that we've worked with so far are all "additive" in that they add columns to the original "left" table
- The diagram shows an INNER JOIN on the id field, showing an additional column being added to the original left table



!["INNER JOIN DIAGRAM 01"](<inner_join_diagram_01.png>) 


```python
SELECT *
FROM left_table
INNER JOIN right_table
ON left_table.id = right_table.id;
```

- Here is some familiar INNER JOIN syntax for this join on the id field
- Let's dive deeper into what it means for this join to be "additive"
- Fields with different names are added to the result set with their original names
- Fields with the same name, such as date, are added to the result set too but notice how we now have duplicate date columns with the same name. This can be changed with aliasing

## Semi join
- A semi join chooses records in the first table where a condition is met in the second table
- Instead of using JOIN or set operators, we can leverage the WHERE clause to specify the records to include
- We'll use the two tables, the left_table and the right_table, in the diagram shown



!["Semi join 01](<semi_join_01.png>)


- A semi join chooses records in the first table where a condition is met in the second table
- More specifically, the semi join will return all values of left_table where values of col1 are in a column we specify, namely col2 in the right_table



!["Semi join 02"](<semi_join_02.png>)



- Records not of interest to the semi join have been faded out
- The final result of our semi join is records corresponding to ids 2 and 3



!["Semi join 03"](<semi_join_03.png>)



- Suppose we're interested in determining the presidents of countries that gained independence before 1800
- We select the fields country, continent, and president

```python
SELECT
    country,
    continent,
    president
FROM presidents;
```

!["Semi join 04"](<semi_join_04.png>)

 
- How do we filter this data for independence year, which is a field in the states table? We'll use a semi join
- Before we knew anything about joins, we knew we could leverage the WHERE clause to filter data
- The query that returns countries with indep_year before 1800 is shown


```python
SELECT country
FROM states
WHERE indep_year < 1800;
```

!["Semi join 05"](<semi_join_05.png>)



- The query returns only Spain and Portugal from our database
- We can use this list of countries as a filter by embedding it in a further WHERE clause
    - This is called a subquery 

```python
SELECT
    president,
    country,
    continent
FROM presidents
WHERE country IN
    (SELECT country
    FROM states
    WHERE indep_year < 1800);
```

- It chooses records in the first table where the country matches the list returned by our subquery 



!["Semi join 05"](<semi_join_05.png>)



- Since Spain does not have a president, only the Portuguese president is listed


## Anti join
- The anti join chooses records in the first table where col1 does NOT match in col2 



!["Anti join 01"](<anti_join_01.png>)



- The final result of our anti join is the records corresponding to ids 1 and 4



!["Anti join 02"](<anti_join_02.png>)



- The final result of our anti join is records corresponding to ids 1 and 4
- How may we adapt our semi join to determine countries in the Americas founded after 1800?
- To change our semi join from before to after 1800, we add NOT before the IN statement

```python
SELECT
    president,
    country,
    continent
FROM presidents
WHERE continent LIKE '%America%'
    AND country NOT IN
        (SELECT country
        FROM states
        WHERE indep_year < 1800);
```

- In addition, we add to our first WHERE clause to filter continents in the Americas
- The result shows presidents of countries in the Americas that gained independence after 1800 



!["Anti join 03"](<anti_join_03.png>)



- Only Chile and Uruguay are returned, not the USA, which gained independence before 1800
- Note again that we are only using the few countries in our database for illustration; this is not a compreensive list of countries in the world

# Self joins
- Self joins are tables joined to themseleves
- They can be used to compare parts of the same table
- Let's recall the prime_ministers table. 
- Suppose all prime ministers are conveniing in summeits of their own continents
- we want to create a new table showing all countries in the same continent as pairs

!["Self join 01"](<self_join_01.png>)

- Self joins don't have dedicated syntax as other joins we have seen do
- We can't just write SELF JOIN in SQL code, for example
- Aliasing is required for a self join
- Let's look at a chunk of INNER JOIN code using the prime_ministers table

```python
SELECT 
    p1.country AS country1,
    p2.country AS country2,
    p1.continent
FROM prime_ministers AS p1
INNER JOIN prime_ministers AS p2
```

- The country column is selected twice, and so is the continent column
- The prime_ministers table is on both the left and the right of the JOIN, making this both a self join and an INNER JOIN
- The vital step here is setting the joining fields which we use to match the table to itself
- For each country, we will find multiple matched countries in teh right table, since we are joining continent
- Each of these matched countries will be returned as pairs
- Since this query will return several records, we use LIMIT to return only the first 10 records

```python
SELECT 
    p1.country AS country1,
    p2.country AS country2,
    p1.continent
FROM prime_ministers AS p1
INNER JOIN prime_ministers AS p2
ON p1.continent = p2.continent
LIMIT 10;
```

- The results are a pairing of each country with every other country in the same continent
- Note that our join also paired countries with themselves, since they too are in the same continent as themselves

!["Self join 02"](<self_join_02.png>)

- We don't want to include these, since a leader from Portugal does not need to meet with themselves, for example
- we can use the not equal operator to exclude records where p1.country and p2.country are identical

```python
SELECT 
    p1.country AS country1,
    p2.country AS country2,
    p1.continent
FROM prime_ministers AS p1
INNER JOIN prime_ministers AS p2
ON p1.continent = p2.continent
    AND p1.country <> p2.country
LIMIT 10;
```

- Here's a look at our final table, showing combinations of countries in our databse that are in the same continent, but excluding records where the two country fields are the same
